## **Python-based College Management System**

```
# College Management System using OOP in Python
# Demonstrates inheritance, abstraction, encapsulation, and polymorphism

from abc import ABC,abstractmethod

# Abstract Base Class for all persons in the college
class Person(ABC):
    def __init__(self,name,age):
        self.name=name
        self.age=age
    @abstractmethod
    def display_info(self):
        pass

# Student class inheriting from Person
class Student(Person):
  def __init__(self,name,age,student_id,course):
    super().__init__(name,age)
    self.__student_id=student_id
    self.course=course
  def get_student_id(self):
    return self.__student
  def display_info(self):
    print(f"Name: {self.name}, Age: {self.age}, Student ID: {self.__student_id}, Course: {self.course}")

 # Professor class inheriting from Person
class Professor(Person):
  def __init__(self,name,age,employee_id,department):
    super().__init__(name,age)
    self.__employee_id=employee_id
    self.department=department
  def get_employee_id(self):
    return self.__employee_id
  def display_info(self):
    print(f"Name: {self.name}, Age: {self.age}, Employee ID: {self.__employee_id}, Department: {self.department}")

# Abstract Course class
class Course(ABC):
  def __init__(self,name,code):
    self.name = name
    self.code = code
    self.branches = []
  @abstractmethod
  def set_branches(self):
    pass
  def display_info(self):
    print(f"Course Name: {self.name}, Course Code: {self.code}")
    for branch in self.branches:
      print(f"  - {branch}")

# Concrete Course subclasses with specific branches
class BTech(Course):
    def __init__(self):
        super().__init__("BTech", "CSE101")
        self.set_branches()

    def set_branches(self):
        self.branches = ["Computer Science", "Electronics", "Mechanical", "Civil"]

class BPharma(Course):
    def __init__(self):
        super().__init__("BPharma", "PHA102")
        self.set_branches()

    def set_branches(self):
        self.branches = ["Pharmacology", "Pharmaceutical Chemistry"]

class LLB(Course):
    def __init__(self):
        super().__init__("LLB", "LAW103")
        self.set_branches()

    def set_branches(self):
        self.branches = ["Criminal Law", "Corporate Law"]

class BBA(Course):
    def __init__(self):
        super().__init__("BBA", "BBA104")
        self.set_branches()

    def set_branches(self):
        self.branches = ["Finance", "Marketing", "HR"]

class BCA(Course):
    def __init__(self):
        super().__init__("BCA", "BCA105")
        self.set_branches()

    def set_branches(self):
        self.branches = ["Software Development", "Cyber Security"]

class MCA(Course):
    def __init__(self):
        super().__init__("MCA", "MCA106")
        self.set_branches()

    def set_branches(self):
        self.branches = ["AI & ML", "Big Data", "Advanced Computing"]

class BSc(Course):
    def __init__(self):
        super().__init__("BSc", "SCI107")
        self.set_branches()

    def set_branches(self):
        self.branches = ["Physics", "Chemistry", "Biology"]

class Diploma(Course):
    def __init__(self):
        super().__init__("Diploma", "DIP108")
        self.set_branches()

    def set_branches(self):
        self.branches = ["Electrical", "Mechanical", "Civil"]

class BCom(Course):
    def __init__(self):
        super().__init__("BCom", "COM109")
        self.set_branches()

    def set_branches(self):
        self.branches = ["Accounting", "Banking"]

class MCom(Course):
    def __init__(self):
        super().__init__("MCom", "COM110")
        self.set_branches()

    def set_branches(self):
        self.branches = ["Finance", "Taxation"]

class MTech(Course):
    def __init__(self):
        super().__init__("MTech", "MTE111")
        self.set_branches()

    def set_branches(self):
        self.branches = ["Computer Science", "Thermal Engineering", "VLSI"]

class Nursing(Course):
    def __init__(self):
        super().__init__("Nursing", "NUR112")
        self.set_branches()

    def set_branches(self):
        self.branches = ["General Nursing", "Pediatric Nursing"]

# College class that holds students, professors, and courses
class College:
    def __init__(self, name,code):
        self.name = name
        self.code = code
        self.students = []
        self.professors = []
        self.courses = []

  # Add a course to the college
    def add_course(self, course):
        self.courses.append(course)

  # Add a student to the college
    def add_student(self, student):
        self.students.append(student)

  # Add a professor to the college
    def add_professor(self, professor):
        self.professors.append(professor)

   # Display all courses offered
    def display_courses(self):
        print(f"\n--- Courses Offered by {self.name} ---")
        for course in self.courses:
            course.display_info()

  # Display all students
    def display_students(self):
        print(f"\n--- Students of {self.name} ---")
        for student in self.students:
            student.display_info()
  # Display all professors
    def display_professors(self):
        print(f"\n--- Professors in {self.name} ---")
        for professor in self.professors:
            professor.display_info()
   # Display full college information
    def display_all(self):
        print(f"\n===== Welcome to {self.name} Management System, Code:{self.code} =====")
        self.display_courses()
        self.display_students()
        self.display_professors()

# --------------------- MAIN PROGRAM ---------------------

# Create college object
college = College("ABC College","COL001")

# Instantiate and add all course objects
courses = [
    BTech(), BPharma(), LLB(), BBA(), BCA(),
    MCA(), BSc(), Diploma(), BCom(), MCom(),
    MTech(), Nursing()
]

for course in courses:
    college.add_course(course)

# Add sample students and professors
college.add_student(Student("Kalyani", 21, "S101", "BTech"))
college.add_student(Student("Pushpa", 22, "S102", "MTech"))
college.add_professor(Professor("Dr. Rao", 45, "P201", "Engineering"))
college.add_professor(Professor("Dr. Neha", 50, "P202", "Nursing"))

# Display everything
college.display_all()
```
